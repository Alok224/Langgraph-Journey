```python
import os
from dotenv import load_dotenv
load_dotenv()

groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(model = "llama-3.1-8b-instant",groq_api_key = groq_api_key)
```

```python
from typing_extensions import TypedDict
from langgraph.graph.message import add_messages,AnyMessage
from typing import Annotated
class state(TypedDict):
    messages:Annotated[list[AnyMessage],add_messages]
```

```python
def superbot(state:state):
    return {"messages":llm.invoke(state["messages"])}
```

```python
from langgraph.graph import StateGraph,START,END
from langgraph.checkpoint.memory import MemorySaver
from IPython.display import display,Image
memory = MemorySaver()
graph = StateGraph(state)

# adding the nodes
graph.add_node("superbot",superbot)

# adding the edges
graph.add_edge(START,"superbot")
graph.add_edge("superbot",END)

# compile the graph
graph_memory = graph.compile(memory)
# visulaize the graph
visaulize_graph = graph_memory.get_graph().draw_mermaid_png()
display(Image(visaulize_graph))
```

```python
from pprint import pprint
config = {"configurable":{"thread_id":"1"}}
messages = graph_memory.invoke({"messages":"Hello, How are you?"},config = config)
for i in messages["messages"]:
    i.pretty_print()
```

```python
config = {"configurable":{"thread_id":"2"}}
for chunk in graph_memory.stream({"messages":"Hii, My name is Alok Goyal. I like myself"},config = config, stream_mode="updates"):
    print(chunk)
```

```python
for chunk in graph_memory.stream({"messages":"I also like Jaipur"},config = config, stream_mode="updates"):
    print(chunk)
```

```python
for chunk in graph_memory.stream({"messages":"Hii, My name is Alok Goyal. I like myself"},config = config, stream_mode="values"):
    pprint(chunk)
```
